```python
# Databricks notebook source
from pyspark.sql import SparkSession, functions as F

spark = SparkSession.builder.getOrCreate()
```

```python

dbutils.widgets.text("source_path", "/Volumes/workspace/ibm/raw_data/sales_source_1500.csv")
source_path = dbutils.widgets.get("source_path")
```

```python
df = (spark.read
      .option("header", True)
      .option("inferSchema", True)
      .csv(source_path))

df_bronze = (df
             .withColumn("ingestion_timestamp", F.current_timestamp())
             .withColumn("source_file", F.input_file_name()))

df_bronze.write.format("delta").mode("overwrite") \
    .saveAsTable("workspace.default.capstone_bronze_sales")

print(f"[BRONZE] Loaded {df_bronze.count()} raw records from {source_path}")
```
